### 9. Save the merged data set tests at your desktop.


```r
# Define merge1 and merge2 using the names vector created earlier
merge1 <- data.frame(names = c("Alberich", "John", "Steve", "Chris"), class = c("A", "B", "A", "B"))
merge2 <- data.frame(names = c("Alberich", "John", "Steve", "Chris"), age = c(20, 21, 22, 20))

# Merge files merge1 and merge2
tests <- merge(merge1, merge2, by = "names")
tests
```

### 10. Use all the methods for adding datasets.

```r
# 1. Horizontal join by key/ID (Merge)
merged_data <- merge(merge1, merge2, by = "names")
merged_data
```

```r
# 2. Binding by columns (Side-by-side, requires same number of rows)
cbind_data <- cbind(merge1, merge2)
cbind_data
```

```r
# 3. Binding by rows (Stacked, requires same column names and structures)
# Create a new dataset of additional students with matching columns to merge1
new_students <- data.frame(names = c("Emma", "Liam"), class = c("A", "B"))

rbind_data <- rbind(merge1, new_students)
rbind_data
```
